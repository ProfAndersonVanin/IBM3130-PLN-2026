# **CÁLCULO DE ESPARSIDADE - MATRIZ BoW**



```
Bag of Words Vetores:
[0, 1, 1, 0, 1, 1, 0, 0, 0, 0, 1]
[1, 0, 0, 1, 1, 1, 1, 0, 1, 1, 1]
[0, 0, 0, 0, 1, 1, 0, 1, 0, 0, 1]
```



````python
# Corpus
corpus = [
    "código Python funciona bem",
    "código Java rápido",
    "Python e Java"
]
````

````python
# Importar bibliotecas e recursos
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
````

````python
# Criar lista vazia para guardar todas as palavras do corpus
todas_palavras=[]
````

````python
# Processar cada frase da lista separadamente e converter para minusculo
for frase in corpus:
  palavras_frase = word_tokenize(frase.lower())
  todas_palavras.extend(palavras_frase)
````

````python
todas_palavras
````

````python
# Extrair vocabulário único (sem repetições)
vocabulario = sorted(set(todas_palavras))
vocabulario
````

````python
# Obter e guadar tamanho do vocabulário
tam_vocab = len(vocabulario)
tam_vocab
````

````python
# Obter a quantidade de documentos do corpus
qtd_documentos = len(corpus)
qtd_documentos
````

````python
# Calcular total de posições da matriz com o vocabulario
total_posicoes = qtd_documentos * tam_vocab
total_posicoes
````

````python
# Montar a matriz BoW
matriz_bow = []
for frase in corpus:
  palavras_frase = word_tokenize(frase.lower())
  vetor_frase = [palavras_frase.count(palavra) for palavra in vocabulario]
  matriz_bow.append(vetor_frase)
````

````python
matriz_bow
````

````python
# Criar os vetores de frequencia para cada frase (documento) com base no vocabulario
````

````python
# visualizar os resultados
print("Vocabulario==> ",vocabulario)
for i, linha in enumerate(matriz_bow):
  print(f"Documento {i+1}: {linha} -> ({corpus[i]})")
````

````python
# Converter a lista para numpy array
import numpy as np
matriz_np = np.array(matriz_bow)
matriz_np
````

````python
total_zeros = np.sum(matriz_np == 0)
total_zeros
````

````python
# Calcular a Esparsidade (%)
esparsidade = total_zeros / total_posicoes
print(f"Esparsidade = {esparsidade*100:.2f}%")
````

````python
from sklearn.feature_extraction.text import CountVectorizer
corpus = [
    "código Python funciona bem",
    "código Java rápido",
    "Python e Java"
]
# Cria um Objeto CountVectorizer
vectorizer = CountVectorizer()
# Treina e transforma o corpus
X = vectorizer.fit_transform(corpus)
# Imprime o vocabulário gerado
print("Vocabulario:", vectorizer.get_feature_names_out())
# Imprime a matriz Bag-of-Words
print("BoW Representação:")
print(X.toarray())
````
