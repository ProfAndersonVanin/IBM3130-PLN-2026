# **Introdução aos Modelos Pré-Treinados com Hugging Face**

## **1. Contextualização: Por que usar Modelos Pré-Treinados?**
Tradicionalmente, treinar modelos de Inteligência Artificial (IA) e Processamento de Linguagem Natural (PLN) exigia recursos colossais: milhões de exemplos de dados, semanas ou meses de processamento em supercomputadores com centenas de GPUs e equipes altamente especializadas.

O cenário mudou radicalmente com a popularização da arquitetura **Transformer** e o conceito de **Aprendizado por Transferência** (**Transfer Learning**):

- **O que é um modelo pré-treinado?** É um modelo que já foi treinado por grandes empresas ou pesquisadores em enormes volumes de dados públicos ou privados. Ele já aprendeu a gramática, o contexto, a semântica da linguagem e padrões complexos.

- **Por que utilizá-los?** Em vez de treinar do zero, você pega esse modelo "já pronto" e o utiliza diretamente para resolver problemas específicos (inferência) ou realiza pequenos ajustes para a sua base de dados (fine-tuning). Isso economiza tempo, energia elétrica e milhões de dólares em custos computacionais.

- **O papel do Hugging Face:** O Hugging Face Hub funciona como o "GitHub da Inteligência Artificial". É um repositório centralizado onde desenvolvedores do mundo todo compartilham modelos, conjuntos de dados (datasets) e métricas prontas para uso.

## **2. A API**

A biblioteca mais famosa do ecossistema é a **transformers**. Para facilitar a vida dos desenvolvedores, ela oferece a função pipeline, uma abstração de alto nível que esconde a complexidade de código por trás de três etapas fundamentais:   
1. **Pré-processamento**: Converte o texto bruto em números que o modelo compreende (tokens).
2. **Modelo**: Executa a inferência matemática através da rede neural.   
3. **Pós-processamento**: Traduz a saída matemática em resultados legíveis por humanos (como "Positivo", "Negativo" ou probabilidades).**

## **Exemplo 1: Análise de Sentimentos (Sentiment Analysis)**
O exemplo mais clássico consiste em analisar se um texto possui uma carga emocional positiva ou negativa. O Hugging Face baixa automaticamente o modelo padrão adequado para essa tarefa na primeira execução.

```python
# Inicializa o pipeline de análise de sentimentos
classificador = pipeline("sentiment-analysis")
```

```python
classificador
```

```python
# Textos de exemplo para teste
textos = [
    "Estou absolutamente apaixonado por essa nova tecnologia de inteligência artificial!",
    "O serviço foi terrivelmente demorado e o produto veio quebrado."
]
```

```python
# Executando o modelo
resultados = classificador(textos)
```

```python
resultados
```

```python
for texto, resultado in zip(textos, resultados):
    print(f"Texto: '{texto}'")
    print(f"Resultado -> Rótulo: {resultado['label']} | Confiança: {resultado['score']:.4f}\n")
```

## **Exemplo 2: Classificação Zero-Shot (Zero-Shot Classification)**
Essa é uma das ferramentas mais poderosas do Hugging Face. Ela permite que um modelo classifique textos em categorias que você inventou na hora, sem que o modelo tenha sido treinado especificamente para elas.

```python
# Cria o pipeline de classificação zero-shot
classificador_zero = pipeline("zero-shot-classification")
```

```python
texto_cliente = "Meu cartão de crédito foi cobrado duas vezes pelo mesmo pedido na loja online."
```

```python
# Definimos rótulos arbitrários que fazemos questão de testar
categorias = ["problemas financeiros", "suporte técnico", "elogio", "cancelamento"]
```

```python
# O modelo avalia qual categoria se encaixa melhor no contexto
resultado = classificador_zero(texto_cliente, candidate_labels=categorias)
```

```python
resultado
```

```python
print(f"Texto analisado: '{texto_cliente}'\n")
for rotulo, pontuacao in zip(resultado['labels'], resultado['scores']):
    print(f"- {rotulo}: {pontuacao:.4f}")
```

## **Exemplo 3: Sumarização de Texto (Summarization)**
Modelos de sumarização leem textos longos (como artigos, relatórios ou notícias) e geram um resumo conciso mantendo os pontos principais.

```python
# Tokenizer
from transformers import T5Tokenizer

# PyTorch model
from transformers import T5Model, T5ForConditionalGeneration
```

```python
token_name = 'unicamp-dl/ptt5-base-portuguese-vocab'
model_name = 'recogna-nlp/ptt5-base-summ'

tokenizer = T5Tokenizer.from_pretrained(token_name )
model_pt = T5ForConditionalGeneration.from_pretrained(model_name)
```

```python
text = '''
“A tendência de queda da taxa de juros no Brasil é real, é visível”, disse Meirelles, que participou na capital americana de uma série de reuniões e encontros com banqueiros e investidores que aconteceram paralelamente às reuniões do Fundo Monetário Internacional (FMI) e do Banco Mundial (Bird) no fim de semana.
Para o presidente do BC, a atual política econômica do governo e a manutenção da taxa de inflação dentro da meta são fatores que garantem queda na taxa de juros a longo prazo.
“Mas é importante que nós não olhemos para isso apenas no curto prazo. Temos que olhar no médio e longo prazos”, disse Meirelles.
Para ele, o trabalho que o Banco Central tem feito para conter a inflação dentro da meta vai gerar queda gradual da taxa de juros.
BC do ano
Neste domingo, Meirelles participou da cerimônia de entrega do prêmio “Banco Central do ano”, oferecido pela revista The Banker à instituição que preside.
“Este é um sinal importante de reconhecimento do nosso trabalho, de que o Brasil está indo na direção correta”, disse ele.
Segundo Meirelles, o Banco Central do Brasil está sendo percebido como uma instituição comprometida com a meta de inflação.
“Isso tem um ganho importante, na medida em que os agentes formadores de preços começam a apostar que a inflação vai estar na meta, que isso é levado a sério no Brasil”, completou.
O presidente do Banco Central disse ainda que a crise política brasileira não foi um assunto de interesse prioritário dos investidores que encontrou no fim de semana.
'''
```

```python
inputs = tokenizer.encode(text, max_length=512, truncation=True, return_tensors='pt')
```

```python
inputs
```

```python
summary_ids = model_pt.generate(inputs, max_length=256, min_length=32, num_beams=5, no_repeat_ngram_size=3, early_stopping=True)
```

```python
summary_ids
```

```python
summary = tokenizer.decode(summary_ids[0])
```

## **Exemplo 4: Tradução Automática (Translation)**
O Hugging Face permite carregar modelos treinados para pares específicos de idiomas. Neste exemplo, usaremos o modelo Helsinki-NLP para traduzir do inglês para o português.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

tokenizer = AutoTokenizer.from_pretrained("unicamp-dl/translation-en-pt-t5")

model = AutoModelForSeq2SeqLM.from_pretrained("unicamp-dl/translation-en-pt-t5")

enpt_pipeline = pipeline('text2text-generation', model=model, tokenizer=tokenizer)

enpt_pipeline("translate English to Portuguese: I like to eat rice.")
```

## **Exemplo 5: Sistema de Perguntas e Respostas (Question Answering)**
Nesta tarefa, fornecemos um contexto (um texto de referência) e fazemos uma pergunta. O modelo localiza exatamente onde está a resposta dentro do texto fornecido.

```python
qa_pipe = pipeline(
    'question-answering',
    model = 'pierreguillou/bert-base-cased-squad-v1.1-portuguese'
)
```

```python
contexto = """
A Petrobras é uma empresa brasileira de capital aberto, com sede no Rio de Janeiro.
Foi fundada em 1953 pelo presidente Getúlio Vargas. É uma das maiores empresas
de energia do mundo, com atuação em exploração e produção de petróleo e gás.
Em 2026, a empresa registrou lucro líquido de R$ 45 bilhões.
O quadro de funcionários tem cerca de 45.000 pessoas.
"""
```

```python
perguntas = [
    'Quando foi fundada a Petrobras?',
    'Quem fundou a Petrobras?',
    'Qual foi o lucro da Petrobras em 2026?',
    'Onde fica a sede da Petrobras?',
    'Quantas pessoas trabalham na Petrobras?',
]
```

```python
for pergunta in perguntas:
    r = qa_pipe(question=pergunta, context=contexto)
    print(f'P: {pergunta}')
    print(f'R: "{r["answer"]}"  (confiança: {r["score"]:.3f})')
    print()
```

## **Exemplo 7:Classificação Zero-Shot**
Este pipeline classifica textos em categorias que você define na hora — sem nenhum dado de treino rotulado.

Internamente, o modelo verifica se cada categoria é uma hipótese plausível para o texto, usando inferência de linguagem natural (NLI).

```python
zs_pipe = pipeline(
    'zero-shot-classification',
    model = 'joeddav/xlm-roberta-large-xnli'
)
```

```python
textos = [
    'O governo anunciou novos investimentos em infraestrutura viária.',
    'A seleção brasileira venceu a Copa América nos pênaltis.',
    'A vacina contra dengue está disponível nos postos de saúde.',
    'A bolsa de valores encerrou em alta de 1,8% nesta quinta-feira.',
    'O novo iPhone tem câmera de 48 megapixels e bateria de longa duração.',
]
```

```python
# Defina aqui as categorias — pode trocar por qualquer coisa!
categorias = ['política', 'esportes', 'saúde', 'tecnologia', 'economia']
```

```python
for texto in textos:
    r = zs_pipe(texto, candidate_labels=categorias)
    #print(f'{texto[:51]:<52} {r["labels"][0]:<12} {r["scores"][0]:.3f}')
    print(f'{texto} {r["labels"][0]} {r["scores"][0]:.3f}')
```
