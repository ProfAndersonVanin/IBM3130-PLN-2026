## 1. O que é POS Tagging?

POS significa **Part of Speech**, ou **classe gramatical**.

POS Tagging é o processo de atribuir uma classe gramatical a cada token de um texto.

Exemplo:

| Palavra | Classe |
|---|---|
| O | DET |
| professor | NOUN |
| prepara | VERB |
| uma | DET |
| aula | NOUN |
| interessante | ADJ |

Em PLN, essa classificação é realizada automaticamente por modelos linguísticos.

## 2. Por que POS Tagging é importante?

POS Tagging pode ser utilizado em:

- análise sintática;
- análise semântica;
- análise de sentimentos;
- extração de informações;
- classificação de textos;
- tradução automática;
- chatbots;
- sistemas de perguntas e respostas.

Por exemplo, em análise de sentimentos, os adjetivos podem fornecer informações importantes sobre a opinião presente em um texto.

## 4. Importando as bibliotecas

````python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from spacy import displacy

nltk.download('punkt')
nltk.download('punkt_tab')
````

## 5. Carregando o modelo de português

````python
nlp = spacy.load("pt_core_news_sm")

print("Modelo carregado com sucesso!")
````

## 6. Criando o primeiro texto

Vamos utilizar uma frase relacionada ao contexto educacional.

````python
texto = "O professor prepara uma aula interessante para os estudantes."

print(texto)
````

## 7. Processando o texto com spaCy

````python
doc = nlp(texto)

print(doc)
````

## 8. Tokenização

````python
for token in doc:
    print(token.text)
````

## 9. POS Tagging

Agora vamos apresentar cada token e sua classe gramatical.

````python
for token in doc:
    print(token.text, "->", token.pos_)
````

## 10. Principais etiquetas POS

| Etiqueta | Significado |
|---|---|
| `NOUN` | Substantivo |
| `PROPN` | Nome próprio |
| `VERB` | Verbo |
| `AUX` | Verbo auxiliar |
| `ADJ` | Adjetivo |
| `ADV` | Advérbio |
| `PRON` | Pronome |
| `DET` | Determinante |
| `ADP` | Preposição |
| `CCONJ` | Conjunção coordenativa |
| `SCONJ` | Conjunção subordinativa |
| `NUM` | Numeral |
| `PUNCT` | Pontuação |

O valor de `token.pos_` fornece uma classificação mais geral e padronizada.

## 11. POS Tagging e lematização

````python
for token in doc:
    print(
        token.text,
        "-> lema:", token.lemma_,
        "| POS:", token.pos_
    )
````

## 12. Comparando `pos_` e `tag_`

`token.pos_` representa uma classe gramatical mais geral.

`token.tag_` pode apresentar uma classificação mais detalhada, dependendo do modelo linguístico.

````python
for token in doc:
    print(
        token.text,
        "| POS:", token.pos_,
        "| TAG:", token.tag_
    )
````

## 15. Criando uma visualização personalizada das POS Tags

Agora vamos adaptar a ideia do tutorial de referência.

Vamos representar as classes gramaticais como elementos destacados no texto.

Primeiro definimos as classes que queremos visualizar.

````python
pos_tags = [
    "NOUN",
    "PROPN",
    "VERB",
    "AUX",
    "ADJ",
    "ADV",
    "PRON",
    "DET",
    "ADP",
    "CCONJ",
    "SCONJ",
    "NUM"
]
````

## 16. Definindo as cores

````python
colors = {
    "NOUN": "turquoise",
    "PROPN": "lightblue",
    "VERB": "lightpink",
    "AUX": "pink",
    "ADJ": "lime",
    "ADV": "orange",
    "PRON": "violet",
    "DET": "lightgreen",
    "ADP": "khaki",
    "CCONJ": "cornflowerblue",
    "SCONJ": "salmon",
    "NUM": "gold"
}
````

## 17. Criando os elementos que serão destacados

Para cada token, precisamos informar:

- posição inicial;
- posição final;
- classe gramatical.

O spaCy fornece a posição inicial por meio de `token.idx`.

````python
ents = []

for token in doc:
    if token.pos_ in pos_tags:
        ents.append({
            "start": token.idx,
            "end": token.idx + len(token.text),
            "label": token.pos_
        })

print(ents)
````

## 18. Criando a estrutura para o displaCy

````python
doc_visual = {
    "text": texto,
    "ents": ents
}

options = {
    "ents": pos_tags,
    "colors": colors
}
````

## 19. Exibindo a visualização personalizada

````python
displacy.render(
    doc_visual,
    style="ent",
    options=options,
    manual=True,
    jupyter=True
)
````

## 20. Criando uma função reutilizável

Vamos transformar todo o processo em uma função chamada `visualizar_pos()`.

````python
def visualizar_pos(texto):

    doc = nlp(texto)

    pos_tags = [
        "NOUN",
        "PROPN",
        "VERB",
        "AUX",
        "ADJ",
        "ADV",
        "PRON",
        "DET",
        "ADP",
        "CCONJ",
        "SCONJ",
        "NUM"
    ]

    colors = {
        "NOUN": "turquoise",
        "PROPN": "lightblue",
        "VERB": "lightpink",
        "AUX": "pink",
        "ADJ": "lime",
        "ADV": "orange",
        "PRON": "violet",
        "DET": "lightgreen",
        "ADP": "khaki",
        "CCONJ": "cornflowerblue",
        "SCONJ": "salmon",
        "NUM": "gold"
    }

    ents = []

    for token in doc:

        if token.pos_ in pos_tags:

            ents.append({
                "start": token.idx,
                "end": token.idx + len(token.text),
                "label": token.pos_
            })

    doc_visual = {
        "text": texto,
        "ents": ents
    }

    options = {
        "ents": pos_tags,
        "colors": colors
    }

    displacy.render(
        doc_visual,
        style="ent",
        options=options,
        manual=True,
        jupyter=True
    )
````

## 21. Testando a função

````python
visualizar_pos(
    "O professor prepara uma aula interessante para os estudantes."
)
````

## 22. Testando outras frases

````python
visualizar_pos(
    "A inteligência artificial transforma a educação."
)
````

````python
visualizar_pos(
    "Os estudantes desenvolveram um projeto inovador."
)
````

````python
visualizar_pos(
    "O computador processa rapidamente grandes quantidades de dados."
)
````

## 23. Encontrando somente substantivos

O POS Tagging permite filtrar tokens de acordo com sua classe.

````python
texto = "Os estudantes desenvolveram um projeto inovador."

doc = nlp(texto)

for token in doc:
    if token.pos_ == "NOUN":
        print(token.text)
````

## 24. Encontrando somente verbos

````python
for token in doc:
    if token.pos_ == "VERB":
        print(token.text)
````

## 25. Encontrando somente adjetivos

````python
for token in doc:
    if token.pos_ == "ADJ":
        print(token.text)
````

## 26. Aplicação em análise de sentimentos

Considere:

> O computador é rápido, eficiente e excelente.

Podemos utilizar POS Tagging para localizar os adjetivos.

````python
texto = "O computador é rápido, eficiente e excelente."

doc = nlp(texto)

for token in doc:
    if token.pos_ == "ADJ":
        print(token.text)
````

## 27. Analisando vários textos

O mesmo procedimento pode ser aplicado a um conjunto de frases.

````python
textos = [
    "O computador é rápido.",
    "O sistema é eficiente.",
    "O programa é simples.",
    "A interface é intuitiva."
]

for texto in textos:

    doc = nlp(texto)

    print("\nTexto:", texto)

    for token in doc:
        print(token.text, "->", token.pos_)
````
