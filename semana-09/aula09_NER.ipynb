# 2. O que é NER?

**Named Entity Recognition (NER)** é uma técnica de Processamento de Linguagem
Natural utilizada para identificar entidades relevantes dentro de um texto.

Exemplo:

> O professor Anderson trabalha em Ribeirão Pires.

Podemos identificar:

| Texto | Tipo |
|---|---|
| Anderson | Pessoa |
| Ribeirão Pires | Local |

O computador analisa o texto e tenta determinar:

**"Esta palavra ou expressão representa o quê?"**

Isso é diferente de simplesmente procurar uma palavra.

Por exemplo, encontrar a palavra "São Paulo" é fácil.

O desafio é descobrir que:

> São Paulo é uma localização geográfica.

O contexto é fundamental.

# **Importação**

````python
import spacy

nlp = spacy.load("pt_core_news_sm")

print("Modelo carregado com sucesso!")
````

# 6. Nosso primeiro exemplo de NER

Vamos utilizar um modelo pré-treinado para português.

O spaCy recebe um texto e realiza várias etapas de processamento.

Depois podemos consultar as entidades encontradas utilizando:

```python
doc.ents

````python
texto = """
Anderson trabalha na IBMEC,
em São Paulo. Ele desenvolve projetos de Inteligência Artificial.
"""

doc = nlp(texto)

for entidade in doc.ents:
    print(entidade.text, "->", entidade.label_)
````

## Entendendo o resultado

Cada entidade possui:

### Texto

A informação encontrada no documento.

### Label

A classificação atribuída pelo modelo.

Por exemplo:

```text
Anderson -> PER
São Paulo -> LOC

A classificação depende do modelo utilizado e pode não ser perfeita.

Por isso, NER é uma tarefa de interpretação automática de linguagem,
e não simplesmente uma busca por palavras.

````python
## Visualização

from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
````

# 7. Explorando as entidades

Podemos consultar algumas informações de cada entidade:

- Texto identificado
- Tipo da entidade
- Descrição do tipo

````python
for entidade in doc.ents:
    print("Texto:", entidade.text)
    print("Tipo:", entidade.label_)
    print("Descrição:", spacy.explain(entidade.label_))
    print()
````

# 8. Criando uma função para reconhecer entidades

Agora vamos transformar nosso código em uma função.

Isso permite reutilizar o mesmo código para diferentes textos.

````python
def reconhecer_entidades(texto):
    doc = nlp(texto)

    for entidade in doc.ents:
        print(entidade.text, "->", entidade.label_)
````

````python
texto = """
A IBMEC em São Paulo, juntamente com o professor Anderson, desenvolve pesquisas
na área de Informática.
"""

reconhecer_entidades(texto)
````

# 9. Encontrando apenas pessoas

Podemos utilizar o tipo da entidade para realizar filtros.

Por exemplo:

```python
entidade.label_ == "PER"

````python
def encontrar_pessoas(texto):
    doc = nlp(texto)

    pessoas = []

    for entidade in doc.ents:
        if entidade.label_ == "PER":
            pessoas.append(entidade.text)

    return pessoas
````

Assim conseguimos encontrar apenas as pessoas identificadas pelo modelo.

````python
texto = """
Anderson trabalha com Cíntia em projetos de Inteligência Artificial.
"""

pessoas = encontrar_pessoas(texto)

print("Pessoas encontradas:")

for pessoa in pessoas:
    print(pessoa)
````

# 10. Organizando as entidades

Podemos criar um dicionário para organizar as entidades encontradas.

Por exemplo:

```text
Pessoas
Organizações
Locais
Datas

# **Extraindo entidades por categoria**

````python
def extrair_entidades(texto):
    doc = nlp(texto)

    resultado = {
        "PESSOAS": [],
        "ORGANIZACOES": [],
        "LOCAIS": [],
        "DATAS": []
    }

    for entidade in doc.ents:

        if entidade.label_ == "PER":
            resultado["PESSOAS"].append(entidade.text)

        elif entidade.label_ == "ORG":
            resultado["ORGANIZACOES"].append(entidade.text)

        elif entidade.label_ in ["LOC", "GPE"]:
            resultado["LOCAIS"].append(entidade.text)

        elif entidade.label_ == "DATE":
            resultado["DATAS"].append(entidade.text)

    return resultado
````

Isso começa a transformar o NER em uma ferramenta de extração estruturada
de informação.

````python
texto = """
Anderson trabalha na Ibmec,
em São Paulo. Em janeiro de 2027, ele participará
de um projeto de Inteligência Artificial.
"""

resultado = extrair_entidades(texto)

resultado
````

### Desafio

O modelo pode apresentar dificuldades para identificar valores
monetários.

Utilize também uma expressão regular (`re`) para localizar
informações que começam com `R$`.

````python
import re

texto = """
A Prefeitura de São Paulo anunciou nesta segunda-feira
um novo programa de tecnologia. O projeto terá investimento
de R$ 50 milhões e será desenvolvido em parceria com
a Universidade de São Paulo. O lançamento está previsto
para março de 2027.
"""

# NER
doc = nlp(texto)

for entidade in doc.ents:
    print(entidade.text, "->", entidade.label_)

print("\nValores encontrados:")

# Expressão regular
valores = re.findall(r"R\$\s?[\d.,]+\s?(?:milhões|mil)?", texto)

for valor in valores:
    print(valor)
````

# Reflexão

O NER consegue identificar automaticamente diversas informações,
mas não é perfeito.

Um modelo pode:

- Não identificar uma entidade.
- Classificar uma entidade incorretamente.
- Confundir uma pessoa com uma organização.
- Ter dificuldades com nomes desconhecidos.
- Ter dificuldades com textos muito específicos.
- Apresentar resultados diferentes dependendo do idioma e do modelo.

Por isso, em aplicações reais, devemos sempre avaliar o desempenho
do modelo.
